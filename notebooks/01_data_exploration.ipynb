```python
# 1. Load dataset
from datasets import load_dataset

dataset = load_dataset("stanfordnlp/imdb")

dataset
```

```python
# 2. Inspect dataset
print(dataset)
print(dataset["train"][0])
```

```python
# 3. Dataset size
print("Training examples:", len(dataset["train"]))
print("Test examples:", len(dataset["test"]))
```

```python
# 4. Label distribution
from collections import Counter

train_labels = Counter(dataset["train"]["label"])
test_labels = Counter(dataset["test"]["label"])

print("Training labels:", train_labels)
print("Test labels:", test_labels)
```

```python
# 5. Review length analysis
import numpy as np

train_lengths = [
    len(example["text"].split())
    for example in dataset["train"]
]

print("Average words:", np.mean(train_lengths))
print("Minimum words:", np.min(train_lengths))
print("Maximum words:", np.max(train_lengths))
print("Median words:", np.median(train_lengths))
```

```python
# 6. Review length visualization
import matplotlib.pyplot as plt

plt.hist(train_lengths, bins=50)
plt.xlabel("Number of words")
plt.ylabel("Number of reviews")
plt.title("IMDb Review Length Distribution")
plt.show()
```
